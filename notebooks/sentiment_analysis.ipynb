# Customer sentiment analysis

This notebook explores the labeled review data and trains the same TF-IDF + Logistic Regression model used by the app. Replace the included illustrative sample with a representative dataset before interpreting results.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent
reviews = pd.read_csv(ROOT / 'data' / 'reviews.csv')
reviews.head()
```

```python
reviews.info()
reviews.isna().sum()
```

```python
import matplotlib.pyplot as plt

label_col = next(c for c in reviews.columns if c.lower() in {'sentiment', 'label', 'class', 'rating', 'score'})
reviews[label_col].value_counts().plot(kind='bar', title='Review labels')
plt.ylabel('Reviews')
plt.tight_layout()
```
